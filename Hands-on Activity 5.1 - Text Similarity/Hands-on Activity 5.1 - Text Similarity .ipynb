Load the necessary libraries

```python
import nltk
import pandas as pd
```

```python
data = pd.read_csv('songdata.csv')
data.head()
```

Find the lyrics by the artist The Beatles

```python
beatles = data[data.artist=='The Beatles']
beatles.head()
```

Check the number of Beatles songs in the data set

Remove characters from the lyrics

```python
import re

newline = lambda x: re.sub('\n', ' ', x) # remove \n
alphanumeric = lambda x: re.sub('\w*\d\w*', ' ', x) # remove alphanumeric words

corpus = beatles.text.map(newline).map(alphanumeric)
corpus.head()
```

Create a document term matrix using Count Vectorizer with the stop words turned on to English

```python
from sklearn.feature_extraction.text import CountVectorizer
          
cv = CountVectorizer(stop_words="english")
X = cv.fit_transform(corpus).toarray()

dt = pd.DataFrame(X, columns=cv.get_feature_names_out()).set_index(beatles.song)
dt.head()
```

Set display options so that the lyrics aren't cut off

```python
pd.set_option('display.max_colwidth', None)
```

```python
beatles[beatles.song=='Imagine'].text
```

Imagine lyrics in Count Vectorizer form


```python
imagine = list(dt.loc['Imagine'])
imagine[:20]
```

Define the cosine similarity calculation

```python
from numpy import dot
from numpy.linalg import norm

cosine = lambda v1, v2: dot(v1, v2) / (norm(v1) * norm(v2))
```

Calculate all similarities and sort by the most similar

```python
similarity = [cosine(imagine, song) for song in X]
sorted(list(zip(similarity, beatles.song)), reverse=True)[1:6]
```

Create the TF-IDF Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer
          
cv_tfidf = TfidfVectorizer(stop_words="english")
X_tfidf = cv_tfidf.fit_transform(corpus).toarray()

dt_tfidf = pd.DataFrame(X_tfidf, columns=cv_tfidf.get_feature_names_out()).set_index(beatles.song)
dt_tfidf.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
          
cv_tfidf = TfidfVectorizer(stop_words="english")
X_tfidf = cv_tfidf.fit_transform(corpus).toarray()

dt_tfidf = pd.DataFrame(X_tfidf, columns=cv_tfidf.get_feature_names_out()).set_index(beatles.song)
dt_tfidf.head()
```

Calculate all similarities and sort by the most similar using the TF-IDF data

```python
similarity_tfidf = [cosine(imagine, song) for song in X_tfidf]
sorted(list(zip(similarity_tfidf, beatles.song)), reverse=True)[1:6]
```

Calculate the cosine similarity between all combinations of documents

```python

from itertools import combinations

pairs = list(combinations(beatles.song.index, 2)) # all song index combos
pairs_0 = list(combinations(range(len(beatles)), 2)) # all index combos starting with (0,1)
song_pairs = [(beatles.song[a_index], beatles.song[b_index]) for (a_index, b_index) in pairs]
song_pairs[:5]
```

 Calculate the cosine similarity of the Count Vectorizer document-term matrix

```python

results = [cosine(X[a_index], X[b_index]) for (a_index, b_index) in pairs_0]
sorted(zip(results, song_pairs), reverse=True)[:5]
```

Calculate the cosine similarity of the TF-IDF Vectorizer document-term matrix

```python

results = [cosine(X_tfidf[a_index], X_tfidf[b_index]) for (a_index, b_index) in pairs_0]
sorted(zip(results, song_pairs), reverse=True)[:5]
```

#### Supplementary Activity

* Use a dataset that consists of text document
* Remove special characters
* Compare the similar document using Count Vector and TF-IDF Vectorizer
* Use the cosine similarity to measure the similarity of text

```python
# Load Dataset

dataset = pd.read_csv('IMDB Dataset.csv')
dataset.head()
```

```python
# Remove Special Characters

br = lambda x: re.sub(r'[^a-zA-Z\s]', ' ', x)
br_2 = lambda x: re.sub(r'br', '', x)

corpus = dataset.review.map(br).map(br_2)

corpus[:25]
```

```python
# Compare the similar document using Count Vector and TF-IDF Vectorizer

# Count Vectorizer
c_vectorizer = CountVectorizer()
X_c_vectorizer = c_vectorizer.fit_transform(corpus)
```

```python
# TF-IDF

tf_idf = TfidfVectorizer()
X_tf_idf = tf_idf.fit_transform(corpus)
```

```python
# Use the cosine similarity to measure the similarity of text
from sklearn.metrics.pairwise import cosine_similarity as cosine

# group by pairs
index_pairs = [(0, i) for i in range(1, 11)]

# Results for Count Vectorizer version
results_cv = [cosine(X_c_vectorizer[a], X_c_vectorizer[b])[0][0] for (a, b) in index_pairs]

# Results for TF-IDF version
results_tf_idf = [cosine(X_tf_idf[a], X_tf_idf[b])[0][0] for (a, b) in index_pairs]
```

```python
# Create DataFrame to view results
df_comparison = pd.DataFrame({
    'Pair': [f"Review 0 vs Review {i}" for i in range(1, 11)],
    'Count Vectorizer': results_cv,
    'TF-IDF': results_tf_idf
})

# Display the table
print(df_comparison)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
#### Tokenize (words)

raw_text = dataset.review[2]
tokens_1 = word_tokenize(raw_text)
print(tokens_1[:100])
```

```python
# Tokenize (words with remove punctuation) / Tokenize (words with remove special characters and numbers)

clean_2 = re.sub(r'[^a-zA-Z\s]', ' ', raw_text)
tokens_2 = word_tokenize(clean_2)
print(tokens_2[:100])
```

```python
# Tokenize (words with remove punctuation and all lowercase)

tokens_3 = [word.lower() for word in tokens_2]
print(tokens_3[:100])
```

```python
# Tokenize (words with remove punctuation, all lowercase and remove stop words)

stop_words = set(stopwords.words('english'))
tokens_4 = [word for word in tokens_3 if word not in stop_words]
print(tokens_4[:100])
```

```python
# Tokenize (words with remove punctuation, all lowercase and  remove stop words, with speech_tagging)

tokens_5 = nltk.pos_tag(tokens_4)
print(tokens_5[:100])
```

#### Conclusion

The Count Vectorizer shows a high similarity of around 60% on average because movie reviews share a common general vocabulary. However, the TF-IDF scores are significantly lower than Count Vectorizer. This is because while the reviews use similar filler words, their actual content and topics are really different. The results shows that TF-IDF is much more effective at distinguishing the unique "thought" of the movie reviews
